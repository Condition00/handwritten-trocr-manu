# Handwritten line recognition

Run cells with existing uv Python kernel. Model accepts single-line image input.

```python
from pathlib import Path
import sys, torch
ROOT = Path.cwd()
sys.path.insert(0, str(ROOT))
print(torch.__version__, torch.cuda.is_available())
```

```python
from PIL import Image
from common import load_trocr
MODEL = 'microsoft/trocr-small-handwritten'
IMAGE = ROOT / 'data/iam_line/sample.png'
processor, model, device = load_trocr(MODEL)
image = Image.open(IMAGE).convert('RGB')
pixels = processor(images=image, return_tensors='pt').pixel_values.to(device)
with torch.inference_mode():
    ids = model.generate(pixels, max_new_tokens=128)
print(processor.batch_decode(ids, skip_special_tokens=True)[0])
```
